# UrbanVibe Demo: obs_filter workflow

This notebook demonstrates how to inspect the current configuration and run the observation filter workflow from Python.

```python
from pathlib import Path

from urbanvibe.config import DEFAULT_CONFIG_PATH, LEGACY_CONFIG_PATH
from urbanvibe.io import DEFAULT_RAW_EXCEL_PATH, LEGACY_EXCEL_PATH

print('Preferred workbook:', DEFAULT_RAW_EXCEL_PATH)
print('Legacy workbook   :', LEGACY_EXCEL_PATH)
print('Preferred config  :', DEFAULT_CONFIG_PATH)
print('Legacy config     :', LEGACY_CONFIG_PATH)

print('\nExists?')
for label, path in [
    ('workbook preferred', DEFAULT_RAW_EXCEL_PATH),
    ('workbook legacy', LEGACY_EXCEL_PATH),
    ('config preferred', DEFAULT_CONFIG_PATH),
    ('config legacy', LEGACY_CONFIG_PATH),
]:
    print(f'- {label:18s}: {path.exists()}')
```
